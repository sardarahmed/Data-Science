```python
url = "https://www.yellowpages.com/los-angeles-ca/electricians"
```

```python
scraper = cloudscraper.create_scraper()  # This handles the Cloudflare challenge
```

```python
response = scraper.get(url)
response
```

```python
soup = BeautifulSoup(response.text, 'html.parser')
```

```python

company = soup.find_all('div', class_='result')

```

```python
length = len(company)
length
```

```python
name=[]
phone_number=[]
address=[]

for i in company:
    n = i.select_one('a.business-name span')
    p = i.find('div', class_='phones phone primary')
    a = i.select_one('div.adr, div.street-address, div.locality')

    name.append(n.text.strip() if n else None)
    phone_number.append(p.text.strip() if p else None)
    address.append(a.text.strip() if a else None)
 

df=pd.DataFrame({'Name':name,
   'Phone Number':phone_number,
   'Address':address,
   })

```

```python
final=pd.DataFrame()
for j in range(1,1001):
  webpage=requests.get('https://www.ambitionbox.com/list-of-companies?page={}'.format(j)).text
  soup=BeautifulSoup(webpage,'lxml')
  company=soup.find_all('div',class_='company-content-wrapper')
  name=[]
  rating=[]
  reviews=[]
  ctype=[]
  hq=[]
  how_old=[]
  no_of_employee=[]

  for i in company:

    try:
       name.append(i.find('h2').text.strip())
    except:
       name.append(np.nan)

    try:
       rating.append(i.find('p',class_='rating').text.strip())
    except:
       rating.append(np.nan)
   
    try:

      reviews.append(i.find('a' , class_='review-count').text.strip())
    except:
      reviews.append(np.nan)

    try:

      ctype.append(i.find_all('p',class_='infoEntity')[0].text.strip())
    except:
      ctype.append(np.nan)
    try:

      hq.append(i.find_all('p',class_='infoEntity')[1].text.strip())
    except:
      hq.append(np.nan)
    
    try:

      how_old.append(i.find_all('p',class_='infoEntity')[2].text.strip())
    except:
      how_old.append(np.nan)
    try:
      no_of_employee.append(i.find_all('p',class_='infoEntity')[3].text.strip())
    except:
      no_of_employee.append(np.nan)
    

  df=pd.DataFrame({'name':name,
    'rating':rating,
    'reviews':reviews,
    'company_type':ctype,
    'Head_Quarters':hq,
    'Company_Age':how_old,
    'No_of_Employee':no_of_employee,
    })
  
  final=final.append(df,ignore_index=True)
```
