```python
import pandas as pd
from bs4 import BeautifulSoup
import cloudscraper

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"
}

Artiles = []
Links = []
Date = []

scraper = cloudscraper.create_scraper()

for p in range(0, 22):
    url = f"https://www.ccdi.gov.cn/scdcn/zggb/zjsc/index_{p}.html"
    response = scraper.get(url, headers=headers)
    response.encoding = 'utf-8'
    soup = BeautifulSoup(response.text, 'html.parser')
    articles = soup.find_all('li')

    for article in articles:
        title = article.select_one('div.title a ')
        link = article.select_one('div.title a')['href']
        date_div = article.select_one('div.more ')
        if date_div:
            date_text = date_div.text.strip()
        else:
            date_text = 'No date available'
        Artiles.append(title.get_text(strip=True))
        Links.append(link)
        Date.append(date_text)

df = pd.DataFrame({
    'Article': Artiles,
    'Link': Links,
    'Date': Date,
})

df.to_csv('articles_shorts.csv', index=False, encoding='utf-8-sig')
```

```python
import pandas as pd
from bs4 import BeautifulSoup
import cloudscraper
import time
import random
from urllib.parse import urljoin

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"
}

Artiles = []
Links = []
Date = []

def scrape_page(url, scraper):
    response = scraper.get(url, headers=headers)
    response.encoding = 'utf-8'
    soup = BeautifulSoup(response.text, 'html.parser')
    count = 0
    for li in soup.select('li:has(div.title)'):
        title_div = li.find('div', class_='title')
        a_tag = title_div.find('a') if title_div else None
        if not a_tag:
            continue
        title = a_tag.get_text(strip=True)
        # Always join with the current page's URL for robustness
        link = urljoin(url, a_tag.get('href', ''))
        date_div = li.find('div', class_='more')
        date_text = date_div.text.strip() if date_div else 'No date available'
        Artiles.append(title)
        Links.append(link)
        Date.append(date_text)
        count += 1
    print(f"Scraped {count} articles from {url}")

start_page = 0
end_page = 4  # Set as needed
batch_size = 2

current = start_page
while current <= end_page:
    # Create a new cloudscraper session for each batch
    scraper = cloudscraper.create_scraper()
    for p in range(current, min(current + batch_size, end_page + 1)):
        if p == 0:
            url = "https://www.ccdi.gov.cn/scdcn/zggb/zjsc/index.html"
        else:
            url = f"https://www.ccdi.gov.cn/scdcn/zggb/zjsc/index_{p}.html"
        scrape_page(url, scraper)
        time.sleep(random.uniform(2, 3))  # Wait 2-5 seconds between pages
    current += batch_size
    print("Waiting 5 seconds before next batch...")
    time.sleep(5)  # Wait 5 seconds

df = pd.DataFrame({
    'Article': Artiles,
    'Link': Links,
    'Date': Date,
})

print(f"Total articles scraped: {len(df)}")
df.to_csv('articles.csv', index=False, encoding='utf-8-sig')
df
```

```python
# aliexpress_scraper.py

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import time
import pandas as pd

def scrape_aliexpress(keyword, max_products=20):
    options = Options()
    options.add_argument("--headless=new")  # Run in background
    options.add_argument('--no-sandbox')
    options.add_argument('--disable-dev-shm-usage')

    driver = webdriver.Chrome(options=options)
    search_url = f"https://www.aliexpress.com/wholesale?SearchText={keyword.replace(' ', '%20')}"
    driver.get(search_url)

    time.sleep(5)  # Wait for page to load

    titles, prices, orders, ratings, urls = [], [], [], [], []

    products = driver.find_elements(By.CSS_SELECTOR, ".manhattan--container--1lP57Ag")[:max_products]

    for product in products:
        try:
            title = product.find_element(By.CSS_SELECTOR, ".manhattan--titleText--WccSjUS").text
            price = product.find_element(By.CSS_SELECTOR, ".manhattan--price-sale--1CCSZfK").text
            order_text = product.find_element(By.CSS_SELECTOR, ".manhattan--trade--2PeJIEi").text
            rating = product.find_element(By.CSS_SELECTOR, ".manhattan--rating--2nUIRY6").text
            link = product.find_element(By.TAG_NAME, 'a').get_attribute("href")

            titles.append(title)
            prices.append(price)
            orders.append(order_text)
            ratings.append(rating)
            urls.append(link)

        except Exception as e:
            continue

    driver.quit()

    df = pd.DataFrame({
        'Title': titles,
        'Price': prices,
        'Orders': orders,
        'Rating': ratings,
        'URL': urls
    })

    df.to_csv("aliexpress_products.csv", index=False)
    print(f"\n✅ {len(df)} products saved to 'aliexpress_products.csv'")

# Run the scraper
if __name__ == "__main__":
    keyword = input("Enter product keyword (e.g., smart LED): ")
    scrape_aliexpress(keyword)
```
