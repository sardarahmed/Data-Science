# Link 2 : https://www.ccdi.gov.cn/scdcn/zyyj/zjsc/index.html

```python
import pandas as pd
from bs4 import BeautifulSoup
import cloudscraper
import time
import random
from urllib.parse import urljoin

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"
}

Artiles = []
Links = []
Date = []

def scrape_page(url, scraper):
    response = scraper.get(url, headers=headers)
    response.encoding = 'utf-8'
    soup = BeautifulSoup(response.text, 'html.parser')
    count = 0
    for li in soup.select('li:has(div.title)'):
        title_div = li.find('div', class_='title')
        a_tag = title_div.find('a') if title_div else None
        if not a_tag:
            continue
        title = a_tag.get_text(strip=True)
        # Always join with the current page's URL for robustness
        link = urljoin(url, a_tag.get('href', ''))
        date_div = li.find('div', class_='more')
        date_text = date_div.text.strip() if date_div else 'No date available'
        Artiles.append(title)
        Links.append(link)
        Date.append(date_text)
        count += 1
    print(f"Scraped {count} articles from {url}")

start_page = 0
end_page = 48  # Set as needed
batch_size = 5

current = start_page
while current <= end_page:
    # Create a new cloudscraper session for each batch
    scraper = cloudscraper.create_scraper()
    for p in range(current, min(current + batch_size, end_page + 1)):
        if p == 0:
            url = "https://www.ccdi.gov.cn/scdcn/zyyj/zjsc/index.html"
        else:
            url = f"https://www.ccdi.gov.cn/scdcn/zyyj/zjsc/index_{p}.html"
        scrape_page(url, scraper)
        time.sleep(random.uniform(2, 5))  # Wait 2-5 seconds between pages
    current += batch_size
    print("Waiting 5 seconds before next batch...")
    time.sleep(5)  # Wait 5 seconds

df = pd.DataFrame({
    'Article': Artiles,
    'Link': Links,
    'Date': Date,
})

print(f"Total articles scraped: {len(df)}")
df.to_csv('Link-2-Articles.csv', index=False, encoding='utf-8-sig')
```
