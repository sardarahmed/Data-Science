```python
from bs4 import BeautifulSoup
import requests
import csv
```

```python
url = "https://www.chocolate.co.uk/collections/all"
```

```python
req = requests.get(url)
soup = BeautifulSoup(req.text, 'html.parser')
```

```python
soup.title
```

```python
import csv
# extract text from all links with specific class
products = soup.find_all("product-item", class_="product-item")  

# save the output to a file
with open("products.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Product", "Price"])
    for product in products:
        title_tag = product.find(class_="product-item-meta__title")
        price_tag = product.find(class_="price")
        
        title = title_tag.get_text(strip=True) if title_tag else "No Title"
        price = price_tag.get_text(strip=True) if price_tag else "No Price"
        
        # Write to CSV file
        writer.writerow([title, price])
```

```python
website = "https://books.toscrape.com/"
req = requests.get(website)
soup = BeautifulSoup(req.text, 'html.parser')
```

```python
with open("books.csv", "w",newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Title"])
    for a in soup.select('article.product_pod h3 a[title]'):
        writer.writerow([a['title']])
```
